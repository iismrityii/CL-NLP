Experiment-3: Stemming, Lemmatization and Regular Expression

Experiment 3.1: Write a Python program to perform
stemming using the Porter Stemmer. Apply stemming to the following words. Display the
original words and their stems


```python
import re
import nltk

nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')

from nltk.stem import PorterStemmer, WordNetLemmatizer

stemmer = PorterStemmer()

words = [
    "Unhappiness",
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "connected",
    "connection",
    "computers"
]

for word in words:
    stem = stemmer.stem(word)
    print(f"{word : <20} {stem}")
```

Experiment 3.2: Write a Python program to perform lemmatization on a given input words:
<br>cats, dogs, running, runs, ran, studies, studying, better, children, mice, went, ate,
leaves, caring,

```python
lemmatizer = WordNetLemmatizer()

words2 = [
    "cats","dog","running","runs","ran","studies", "studying", "better", "children", "mice", "went", "ate", "leaves", "caring"
]

for word in words2:
    lemma = lemmatizer.lemmatize(word)
    print(f"{word:<20} {lemma}")
```

Experiment 3.3: Write a Python program using Regular Expressions (Regex) to extract
and process the following information from a text:

1) Email addresses
2) URLs
3) Mobile numbers
4) Hashtags
5) Mentions

<br><br>
Input Text:<br><br>
Welcome to the Natural Language Processing workshop! For more information, visit
https://www.nlpworkshop.com or www.python.org. You can contact the coordinator at
nlpworkshop@gmail.com or support@python.org. For registration, call +91-9876543210 or
9123456789. Follow us on social media @NLPWorkshop and @PythonLearner. Share your
experience using #NLP, #Python, and #MachineLearning. You can also visit

```python
text = """
Welcome to the Natural Language Processing workshop! For more information, visit
https://www.nlpworkshop.com or www.python.org. You can contact the coordinator at
nlpworkshop@gmail.com or support@python.org. For registration, call +91-9876543210 or
9123456789. Follow us on social media @NLPWorkshop and @PythonLearner. Share your
experience using #NLP, #Python, and #MachineLearning. You can also visit
https://github.com/NLPWorkshop for the latest updates.
"""

email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'

emails = re.findall(email_pattern, text)

print("\n1. Emails")


for email in emails:
    print(email)

url_pattern = r'https?://[^\s]+|www\.[^\s]+'

urls = re.findall(url_pattern, text)

print("\n2. URLs")
# print("-" * 30)

for url in urls:
    print(url)

mobile_pattern = r'(?:\+91[-\s]?)?[6-9]\d{9}'

mobile_numbers = re.findall(mobile_pattern, text)

print("\n3. MOBILE NUMBERS")
# print("-" * 30)

for number in mobile_numbers:
    print(number)


hashtag_pattern = r'#[A-Za-z0-9_]+'

hashtags = re.findall(hashtag_pattern, text)

print("\n4. HASHTAGS")
# print("-" * 30)

for hashtag in hashtags:
    print(hashtag)

mention_pattern = r'@[A-Za-z0-9_]+'

mentions = re.findall(mention_pattern, text)

print("\n5. MENTIONS")
# print("-" * 30)

for mention in mentions:
    print(mention)

print("\nEmails:", emails)
print("URLs:", urls)
print("Mobile Numbers:", mobile_numbers)
print("Hashtags:", hashtags)
print("Mentions:", mentions)
```
