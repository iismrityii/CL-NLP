### **Experiment 2.1: Text Preprocessing**

This experiment aims to demonstrate basic text preprocessing steps on a given text file. We will perform the following operations:

1.  **Read Text Data**: Load the content from `/content/2.1_text_data.txt`.
2.  **Convert to Lowercase**: Transform all text to lowercase.
3.  **Remove Punctuation**: Eliminate all punctuation marks.
4.  **Remove Numbers**: Remove all numeric characters.
5.  **Remove Extra Whitespace**: Standardize spacing to single spaces.

For each step, we will also display relevant statistics, such as counts of uppercase letters, types of punctuation, numbers, and extra whitespace, before applying the transformations.

```python
import re

file_path = '/content/2.1_text_data.txt'

with open(file_path, 'r') as file:
    original_text = file.read()

print("Original Text:\n")
print(original_text)
print(f"\nLength of original text: {len(original_text)} characters")
```

### Convert text to lowercase and count uppercase characters

```python
uppercase_chars = sum(1 for char in original_text if char.isupper())
print(f"Number of uppercase characters: {uppercase_chars}")

cleaned_text = original_text.lower()

print("\nText after converting to lowercase:\n")
print(cleaned_text)
```

### Remove punctuation and count different types of punctuation

```python
import string

punctuation_found = {}
for char in cleaned_text:
    if char in string.punctuation:
        punctuation_found[char] = punctuation_found.get(char, 0) + 1

print(f"Different types of punctuations and their counts: {punctuation_found}")
print(f"Total number of punctuation characters: {sum(punctuation_found.values())}")

cleaned_text = cleaned_text.translate(str.maketrans('', '', string.punctuation))

print("\nText after removing punctuation:\n")
print(cleaned_text)
```

### Remove numbers and count them

```python
numbers_found = sum(1 for char in cleaned_text if char.isdigit())
print(f"Number of numeric characters: {numbers_found}")

cleaned_text = re.sub(r'\d+', '', cleaned_text)

print("\nText after removing numbers:\n")
print(cleaned_text)
```

### Remove extra whitespace and count how much extra whitespace is present

```python
extra_whitespace_count = 0
extra_whitespace_count += sum(len(match.group(0)) - 1 for match in re.finditer(r'\s{2,}', cleaned_text))

leading_spaces_match = re.match(r'^\s*', cleaned_text)
trailing_spaces_match = re.search(r'\s*$', cleaned_text)
extra_whitespace_count += len(leading_spaces_match.group(0)) if leading_spaces_match else 0
extra_whitespace_count += len(trailing_spaces_match.group(0)) if trailing_spaces_match else 0

print(f"Amount of extra whitespace present (characters): {extra_whitespace_count}")

cleaned_text = re.sub(r'\s+', ' ', cleaned_text).strip()

print("\nText after removing extra whitespace:\n")
print(cleaned_text)
print(f"\nLength of cleaned text: {len(cleaned_text)} characters")
```

### **Experiment 2.2: Word Tokenization and Sentence Tokenization**

This experiment focuses on breaking down text into individual words (word tokenization) and sentences (sentence tokenization) using different approaches:

1.  **NLTK (Natural Language Toolkit)**: A popular library for natural language processing.
2.  **spaCy**: An industrial-strength NLP library.
3.  **Custom Implementation**: Tokenization without external NLP libraries.

```python
file_path_exp2_2 = '/content/2.2_tokenization_data.txt'

with open(file_path_exp2_2, 'r') as file:
    exp2_2_text = file.read()

print("Text for Tokenization:\n")
print(exp2_2_text)
```

#### **Using NLTK for Tokenization**

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize

try:
    nltk.data.find('tokenizers/punkt_tab')
except LookupError:
    print("Downloading NLTK 'punkt_tab' tokenizer...")
    nltk.download('punkt_tab', quiet=True)

nltk_words = word_tokenize(exp2_2_text)
print("\nNLTK Word Tokens:\n", nltk_words)

nltk_sentences = sent_tokenize(exp2_2_text)
print("\nNLTK Sentence Tokens:\n", nltk_sentences)
```

#### **Using spaCy for Tokenization**

```python
import spacy

try:
    nlp = spacy.load("en_core_web_sm")
except OSError:
    print("Downloading spaCy model 'en_core_web_sm'...")
    spacy.cli.download("en_core_web_sm")
    nlp = spacy.load("en_core_web_sm")

doc = nlp(exp2_2_text)

spacy_words = [token.text for token in doc]
print("\nspaCy Word Tokens:\n", spacy_words)

spacy_sentences = [sent.text for sent in doc.sents]
print("\nspaCy Sentence Tokens:\n", spacy_sentences)
```

#### **Tokenization Without NLTK/spaCy**

```python
import re
custom_words = re.findall(r'\b\w+\b', exp2_2_text.lower())
print("\nCustom Word Tokens:\n", custom_words)

custom_sentences = re.split(r'(?<=[.!?])\s+', exp2_2_text)
print("\nCustom Sentence Tokens:\n", custom_sentences)
```

### **Experiment 2.3: Stop Word Removal**

This experiment demonstrates how to remove commonly occurring stop words from a given text. Stop words are words (like 'is', 'the', 'a') that do not carry much meaning and are often removed in NLP tasks to reduce noise.

```python
file_path_exp2_3 = '/content/2.3_clean_data.txt'

with open(file_path_exp2_3, 'r') as file:
    exp2_3_text = file.read()

print("Text for Stop Word Removal:\n")
print(exp2_3_text)
```

#### **Tokenize and Remove Stop Words**

```python
import re
stop_words = set(["i", "me", "my", "myself", "we", "our", "ours", "ourselves", "you", "your", "yours", "yourself", "yourselves", "he", "him", "his", "himself", "she", "her", "hers", "herself", "it", "its", "itself", "they", "them", "their", "theirs", "themselves", "what", "which", "who", "whom", "this", "that", "these", "those", "am", "is", "are", "was", "were", "be", "been", "being", "have", "has", "had", "having", "do", "does", "did", "doing", "a", "an", "the", "and", "but", "if", "or", "because", "as", "until", "while", "of", "at", "by", "for", "with", "about", "against", "between", "into", "through", "during", "before", "after", "above", "below", "to", "from", "up", "down", "in", "out", "on", "off", "over", "under", "again", "further", "then", "once", "here", "there", "when", "where", "why", "how", "all", "any", "both", "each", "few", "more", "most", "other", "some", "such", "no", "nor", "not", "only", "own", "same", "so", "than", "too", "very", "s", "t", "can", "will", "just", "don", "should", "now"])

words = re.findall(r'\b\w+\b', exp2_3_text.lower())

print("Original Tokens:\n", words)

filtered_tokens = [word for word in words if word not in stop_words]

print("\nStop Words Found and Removed:\n", [word for word in words if word in stop_words])
print("\nFiltered Tokens (without stop words):\n", filtered_tokens)
```
