4.1. To perform Term-Frequency Analysis and Named Entity Recognition on a given text using
any toolkit.

● Calculate the frequency of each word and save the output in CSV format (Term,
Frequency)

● Identify the top 10 most frequent terms and display their corresponding
frequencies.

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab') 
nltk.download('words')
```

```python
import re
import nltk
import csv
from collections import Counter


with open("4.1_4.2_input.txt", 'r', encoding="utf-8") as file:
    text = file.read()

words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

frequency = Counter(words)

for word, count in frequency.most_common():
    print(f"{word:<20} {count}")

with open("term_frequency.csv", 'w', newline='', encoding="utf-8") as file:
    writer =csv.writer(file)

    writer.writerow(["Term", "Frequency"])

    for word, count in frequency.most_common():
        writer.writerow([word, count])

print("\nCSV File Created at!")
```

```python
for i, (word, count) in enumerate(frequency.most_common(10), start = 1):
    print(f"{i}. {word} -> {count}")

print("\nNamed Entity Recognition")

tokens = nltk.word_tokenize(text)
tagged_words = nltk.pos_tag(tokens)
named_entities = nltk.ne_chunk(tagged_words)

for entity in named_entities:
    if hasattr(entity, "label"):
        entity_name = " ".join(word for word, tag in entity.leaves())
        print(f"{entity_name} -> {entity.label()}")
```

4.2. To perform Term-Frequency Analysis on a given text without using any toolkit.

● Calculate the frequency of each word and save the output in CSV format (Term,
Frequency)

● Identify the top 10 most frequent terms and display their corresponding
frequencies

```python
import re
import csv
from collections import Counter


with open("4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()


text = text.lower()


text = re.sub(r'[^\w\s]', '', text)
words = text.split()

frequency = Counter(words)


print("TERM FREQUENCY ANALYSIS")

print("\nTerm\t\tFrequency")


for word, count in frequency.most_common():
    print(f"{word:<20}{count}")

with open("term_frequency_4_2.csv", "w", newline="", encoding="utf-8") as file:

    writer = csv.writer(file)

    writer.writerow(["Term", "Frequency"])

    for word, count in frequency.most_common():
        writer.writerow([word, count])

print("\nCSV file created:")
print("term_frequency_4_2.csv")

print("\n")
# print("=" * 60)
print("TOP 10 MOST FREQUENT TERMS")
# print("=" * 60)

top_10 = frequency.most_common(10)

for i, (word, count) in enumerate(top_10, start=1):

    print(f"{i}. {word} -> {count}")
```

4.3. Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:

1. Read multiple text documents.

2. Convert text to lowercase.

3. Remove punctuation.

4. Tokenize the documents.

5. Calculate TF for every term.

6. Calculate DF for every term.

7. Calculate IDF using the logarithmic formula.

8. Calculate TF-IDF for every term in every document.

9. Display Term, TF, DF, IDF, and TF-IDF at each step.

10. Identify the top 10 terms with the highest TF-IDF score for each document.

Given Input

Consider the following three documents:

Document 1:

Natural language processing is a field of artificial intelligence.

Document 2:

Natural language processing helps computers understand human language.

Document 3:

Machine learning is an important part of artificial intelligence

```python
import re
import math
from collections import Counter


documents = {

    "Document 1":
        "Natural language processing is a field of artificial intelligence.",

    "Document 2":
        "Natural language processing helps computers understand human language.",

    "Document 3":
        "Machine learning is an important part of artificial intelligence."
}



processed_documents = {}

for document_name, text in documents.items():

    # Convert to lowercase
    text = text.lower()

    # Remove punctuation
    text = re.sub(r'[^\w\s]', '', text)

    # Tokenize
    words = text.split()

    processed_documents[document_name] = words



# print("=" * 70)
print("PROCESSED DOCUMENTS")
# print("=" * 70)

for document, words in processed_documents.items():

    print(f"\n{document}:")
    print(words)


print("\n")
# print("=" * 70)
print("TERM FREQUENCY (TF)")
# print("=" * 70)

tf = {}

for document, words in processed_documents.items():

    word_count = Counter(words)

    total_words = len(words)

    tf[document] = {}

    print(f"\n{document}")
    print("-" * 50)

    for word, count in sorted(word_count.items()):

        tf_value = count / total_words

        tf[document][word] = tf_value

        print(
            f"Term: {word:<20}"
            f"Count: {count:<5}"
            f"TF: {tf_value:.4f}"
        )




print("\n")
print("=" * 70)
print("DOCUMENT FREQUENCY (DF)")
print("=" * 70)


# Create vocabulary containing every unique term
vocabulary = set()

for words in processed_documents.values():

    vocabulary.update(words)


df = {}

for term in sorted(vocabulary):

    count = 0

    for words in processed_documents.values():

        if term in words:
            count += 1

    df[term] = count


print("\nTerm\t\t\tDF")
print("-" * 40)

for term, value in df.items():

    print(f"{term:<25}{value}")




print("\n")
print("=" * 70)
print("INVERSE DOCUMENT FREQUENCY (IDF)")
print("=" * 70)


number_of_documents = len(processed_documents)

idf = {}

for term in sorted(vocabulary):

    # IDF formula
    idf_value = math.log10(number_of_documents / df[term])

    idf[term] = idf_value

    print(
        f"Term: {term:<25}"
        f"DF: {df[term]:<5}"
        f"IDF: {idf_value:.4f}"
    )




print("\n")
print("=" * 70)
print("TF-IDF")
print("=" * 70)


tf_idf = {}

for document in processed_documents:

    tf_idf[document] = {}

    print(f"\n{document}")
    print("-" * 70)

    for term in sorted(vocabulary):

        # If term exists in document
        if term in tf[document]:

            tf_value = tf[document][term]

        else:

            tf_value = 0

        idf_value = idf[term]

        # TF-IDF formula
        tfidf_value = tf_value * idf_value

        tf_idf[document][term] = tfidf_value

        print(
            f"Term: {term:<20}"
            f"TF: {tf_value:.4f}\t"
            f"DF: {df[term]}\t"
            f"IDF: {idf_value:.4f}\t"
            f"TF-IDF: {tfidf_value:.4f}"
        )

print("\n")
print("=" * 70)
print("TOP 10 TERMS WITH HIGHEST TF-IDF")
print("=" * 70)


for document in tf_idf:

    print(f"\n{document}")
    print("-" * 50)

    # Sort terms according to TF-IDF
    sorted_terms = sorted(
        tf_idf[document].items(),
        key=lambda x: x[1],
        reverse=True
    )

    # Display top 10
    for i, (term, score) in enumerate(sorted_terms[:10], start=1):

        print(f"{i}. {term:<20} {score:.4f}")
```
