5.1. Implement subword-level tokenization using a BPE-based tokenizer. Given
input data:

https://drive.google.com/file/d/1eHA0Q9ju-08n_hByCK03GqWAX9HQvAKd/view?u
sp=sharing

1. Tokenize the sentence into subword units.

2. Display the generated tokens and their corresponding token IDs.

(Implement two separate codes: a. Using pretrained model and b. Without using
pretrained model)

```python
from transformers import GPT2Tokenizer

with open("D:\SEM 7\CLNLP\LAB\Experiment_5\input_sub_word_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

tokens = tokenizer.tokenize(text)

token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("BPE tokens and token ids: \n")

for token, token_id in zip(tokens, token_ids):
    print(f"{token:30} {token_id}")
```

Without using Pretrained Transformer


```python
from tokenizers import Tokenizer
from tokenizers.models import BPE
from tokenizers.trainers import BpeTrainer
from tokenizers.pre_tokenizers import Whitespace

tokenizer = Tokenizer(BPE(unk_token = "[UNK]"))
tokenizer.pre_tokenizer = Whitespace()

trainer = BpeTrainer(
    vocab_size=500,
    min_frequency=2,
    special_tokens=["[UNK]"]
)

tokenizer.train_from_iterator([text], trainer)

output = tokenizer.encode(text)

tokens = output.tokens
token_ids = output.ids

print("Custom BPE token and Token IDS: \n")

for token, token_id in zip(tokens, token_ids):
    print(f"{token:30}{token_id}")

```

```python
print("\n vocabulary Size: ", tokenizer.get_vocab_size())
```

```python
from transformers import T5Tokenizer

with open("input_sub_word_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

tokenizer = T5Tokenizer.from_pretrained("t5-small")

tokens = tokenizer.tokenize(text)

token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("SentencePiece Tokens and Token IDs:\n")

for token, token_id in zip(tokens, token_ids):
    print(f"{token:30} {token_id}")
```

Without using a pretrained model

```python
import sentencepiece as spm
import os

input_file = "input_sub_word_data.txt"
model_prefix = "my_sentencepiece"

# Train a new SentencePiece model
spm.SentencePieceTrainer.train(
    input=input_file,
    model_prefix=model_prefix,
    vocab_size=300,
    model_type="unigram",
    character_coverage=1.0,
    unk_id=0,
    pad_id=1,
    bos_id=2,
    eos_id=3
)

# Load the newly trained model
sp = spm.SentencePieceProcessor(
    model_file=model_prefix + ".model"
)

# Read input text
with open(input_file, "r", encoding="utf-8") as file:
    text = file.read()

# Tokenize
tokens = sp.encode(text, out_type=str)

# Get token IDs
token_ids = sp.encode(text, out_type=int)

# Display tokens and IDs
print("Custom SentencePiece Tokens and Token IDs:\n")

for token, token_id in zip(tokens, token_ids):
    print(f"{token:30} {token_id}")

print("\nVocabulary Size:", sp.get_piece_size())
```

5.3. Implement Part-of-Speech (POS) tagging on a given text and display the unique
token, POS tag, and description of each POS tag.

"The young student is reading an interesting book in the library."

(Implement two separate codes: a. Using Spacy and b. NLTK )

a. using Spacy

```python
import spacy

# Load English language model
nlp = spacy.load("en_core_web_sm")

# Input sentence
text = "The young student is reading an interesting book in the library."

# Process text
doc = nlp(text)

# Store unique tokens
seen = set()

print("Token\t\tPOS\tDescription")
print("-" * 50)

for token in doc:
    if token.text not in seen and not token.is_punct:
        seen.add(token.text)

        print(
            f"{token.text:15}"
            f"{token.pos_:10}"
            f"{spacy.explain(token.pos_)}"
        )
```

b. Using NLTK

```python
import nltk

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")

from nltk import word_tokenize, pos_tag

# Input sentence
text = "The young student is reading an interesting book in the library."

# Tokenize
tokens = word_tokenize(text)

# POS tagging
tagged_tokens = pos_tag(tokens)

# Dictionary containing POS descriptions
pos_descriptions = {
    "CC": "Coordinating conjunction",
    "CD": "Cardinal number",
    "DT": "Determiner",
    "EX": "Existential there",
    "FW": "Foreign word",
    "IN": "Preposition or subordinating conjunction",
    "JJ": "Adjective",
    "JJR": "Adjective, comparative",
    "JJS": "Adjective, superlative",
    "LS": "List item marker",
    "MD": "Modal",
    "NN": "Noun, singular",
    "NNS": "Noun, plural",
    "NNP": "Proper noun, singular",
    "NNPS": "Proper noun, plural",
    "PDT": "Predeterminer",
    "POS": "Possessive ending",
    "PRP": "Personal pronoun",
    "PRP$": "Possessive pronoun",
    "RB": "Adverb",
    "RBR": "Adverb, comparative",
    "RBS": "Adverb, superlative",
    "RP": "Particle",
    "TO": "To",
    "UH": "Interjection",
    "VB": "Verb, base form",
    "VBD": "Verb, past tense",
    "VBG": "Verb, gerund/present participle",
    "VBN": "Verb, past participle",
    "VBP": "Verb, non-3rd person singular present",
    "VBZ": "Verb, 3rd person singular present",
    "WDT": "Wh-determiner",
    "WP": "Wh-pronoun",
    "WP$": "Possessive wh-pronoun",
    "WRB": "Wh-adverb"
}

# Display unique tokens
seen = set()

print("Token\t\tPOS\tDescription")
print("-" * 70)

for token, tag in tagged_tokens:
    if token not in seen and token.isalpha():
        seen.add(token)

        description = pos_descriptions.get(
            tag,
            "Unknown POS tag"
        )

        print(
            f"{token:15}"
            f"{tag:10}"
            f"{description}"
        )
```

5.4. Implement Part-of-Speech (POS) tagging on a given text and display the unique
token, POS tag, description of each token and frequency using Spacy.

Input data: Same input data of 5.1

```python
import spacy
from collections import Counter

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# Read input data
with open("input_sub_word_data.txt", "r", encoding="utf-8") as file:
    text = file.read()

# Process text
doc = nlp(text)

# Count token frequencies
frequency = Counter()

# Store POS information
token_info = {}

for token in doc:

    # Ignore punctuation and spaces
    if token.is_punct or token.is_space:
        continue

    # Convert to lowercase for unique-token counting
    word = token.text.lower()

    # Frequency
    frequency[word] += 1

    # Store POS information
    if word not in token_info:
        token_info[word] = (
            token.pos_,
            spacy.explain(token.pos_)
        )

# Display result
print(
    f"{'Token':20}"
    f"{'POS':10}"
    f"{'Description':45}"
    f"{'Frequency':10}"
)

print("-" * 90)

for token in sorted(token_info.keys()):

    pos, description = token_info[token]

    print(
        f"{token:20}"
        f"{pos:10}"
        f"{description:45}"
        f"{frequency[token]:10}"
    )
```
