### Experiment 1.1: Read the employee dataset and display the average salary of each department using a bar chart.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/employee_information_100.csv')

print("First 5 rows of the dataset:")
display(df.head())
print("\nDataset Info:")
df.info()
```

```python
avg_salary_by_department = df.groupby('Department')['Salary'].mean().reset_index()

print("\nAverage Salary by Department:")
display(avg_salary_by_department)
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Department', y='Salary', data=avg_salary_by_department, palette='viridis')
plt.title('Average Salary of Each Department')
plt.xlabel('Department')
plt.ylabel('Average Salary')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Experiment 1.2: Display the number of employees working in each department.

```python
employees_by_department = df['Department'].value_counts().reset_index()
employees_by_department.columns = ['Department', 'Number of Employees']

print("\nNumber of Employees in Each Department:")
display(employees_by_department)
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(x='Department', y='Number of Employees', data=employees_by_department, palette='viridis')
plt.title('Number of Employees in Each Department')
plt.xlabel('Department')
plt.ylabel('Number of Employees')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

### Experiment 1.3: Display the percentage of male and female employees using a pie chart.

```python
gender_counts = df['Gender'].value_counts()
gender_percentages = gender_counts / len(df) * 100

print("\nPercentage of Male and Female Employees:")
display(gender_percentages.to_frame(name='Percentage'))
```

```python
plt.figure(figsize=(8, 8))
plt.pie(gender_percentages, labels=gender_percentages.index, autopct='%1.1f%%', startangle=140, colors=['skyblue', 'lightcoral'])
plt.title('Percentage of Male and Female Employees')
plt.axis('equal')
plt.show()
```

### Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.

```python
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Experience_Years', y='Salary', data=df, hue='Department', palette='viridis', s=100, alpha=0.7)
plt.title('Relationship between Experience and Salary')
plt.xlabel('Experience (Years)')
plt.ylabel('Salary')
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

### Experiment 1.6: Display the top 10 employees with the highest salary.

```python
top_10_employees = df.sort_values(by='Salary', ascending=False).head(10)

print("\nTop 10 Employees with the Highest Salary:")
display(top_10_employees)
```

### Experiment 1.7: Find the highest salary in every department.

```python
highest_salary_by_department = df.groupby('Department')['Salary'].max().reset_index()

print("\nHighest Salary in Each Department:")
display(highest_salary_by_department)
```

### Experiment 1.8: Display employees whose salary is greater than the overall average salary.

```python
overall_average_salary = df['Salary'].mean()

print(f"\nOverall Average Salary: {overall_average_salary:.2f}")

employees_above_average_salary = df[df['Salary'] > overall_average_salary]

print("\nEmployees with Salary Greater Than Overall Average:")
display(employees_above_average_salary)
```

### Experiment 1.9: Calculate the average years of experience in each department.

```python
avg_experience_by_department = df.groupby('Department')['Experience_Years'].mean().reset_index()

print("\nAverage Years of Experience by Department:")
display(avg_experience_by_department)
```

### Experiment 1.10: Display the distribution of employee ages using a histogram.

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Age'], bins=10, kde=True, color='lightcoral')
plt.title('Distribution of Employee Ages')
plt.xlabel('Age')
plt.ylabel('Number of Employees')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Salary'], bins=10, kde=True, color='skyblue')
plt.title('Salary Distribution of Employees')
plt.xlabel('Salary')
plt.ylabel('Number of Employees')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
